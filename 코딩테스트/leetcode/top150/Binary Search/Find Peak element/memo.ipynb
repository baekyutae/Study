1. 가운데 mid를 잡는다.

2. nums[mid] < nums[mid + 1] 이면
   mid에서 mid+1 방향으로 오르막이다.
   따라서 오른쪽 구간에는 peak가 반드시 존재한다.
   mid는 peak가 아니므로 버려도 된다.
   => left = mid + 1

3. nums[mid] > nums[mid + 1] 이면
   mid에서 mid+1 방향으로 내리막이다.
   따라서 mid를 포함한 왼쪽 구간에는 peak가 반드시 존재한다.
   mid가 peak일 수도 있으므로 버리면 안 된다.
   => right = mid

4. 항상 nums[mid + 1]을 비교해야 하므로
   mid + 1이 유효해야 한다.
   그래서 while 조건은 left < right 를 쓴다.

5. left == right가 되면
   peak가 존재하는 구간이 한 칸으로 좁혀진 것이므로
   그 인덱스를 반환한다.
