1. **한 줄 진단**

* `target을 찾는 이진 탐색`까지는 갔지만, 이 문제의 본질인 `target 구간의 왼쪽 경계 / 오른쪽 경계 찾기`로 목표를 재정의하지 못했다.

---

2. **내 사고가 막힌 지점**

* **처음 접근:**
  정렬 배열에서 `target`을 이진 탐색으로 찾으려고 했다.

* **맞았던 생각:**
  `target`이 여러 개라면 정렬되어 있으므로 연속된 구간으로 존재한다는 생각은 맞았다.
  즉, 답이 “target이 존재하는 index의 최솟값과 최댓값”이라는 해석도 맞았다.

* **틀렸던 생각:**
  `nums[mid] == target`을 만났을 때 “일단 찾음”에서 멈췄다.
  이 문제에서는 찾는 순간이 끝이 아니라, **그 위치보다 더 왼쪽에도 target이 있는지 / 더 오른쪽에도 target이 있는지** 계속 좁혀야 한다.

* **결정적으로 막힌 이유:**
  일반 이진 탐색의 종료 조건은 `target 존재 여부`인데, 이 문제의 종료 조건은 `target이 처음 등장하는 위치`와 `마지막 등장하는 위치`다.
  즉, “값 찾기”가 아니라 “경계 찾기” 문제로 바꿔야 했는데, 그 불변식을 세우지 못했다.

---

3. **실패 유형 / 등급**

* **주 실패 유형:**

  * 상태·불변식 설계 실패

* **등급:**

  * C 해설 후 백지 구현 가능

* **근거:**
  정렬 배열, 이진 탐색, target이 연속 구간이라는 핵심 신호는 잡았다.
  다만 `nums[mid] == target`일 때 `return`하거나 멈추는 사고가 남아 있어서, `정답 후보를 저장하고 탐색 범위를 계속 줄인다`는 경계 탐색 패턴까지는 아직 연결되지 않았다.

---

4. **일반화해서 공부할 요소**

* **알고리즘/자료구조 신호:**
  `sorted array + O(log n) + first/last position`이면 단순 이진 탐색이 아니라 **lower bound / upper bound 계열**이다. 문제는 정렬 배열에서 시작/끝 위치를 찾고, 없으면 `[-1, -1]`을 반환하라고 요구한다. ([leetcode.com][1])

* **상태/불변식:**
  왼쪽 경계를 찾을 때는
  `target을 찾았어도 더 왼쪽에 답이 있을 수 있다`
  → `answer = mid`, `right = mid - 1`

  오른쪽 경계를 찾을 때는
  `target을 찾았어도 더 오른쪽에 답이 있을 수 있다`
  → `answer = mid`, `left = mid + 1`

* **파이썬 문법/구현 포인트:**
  `result = []`로 시작해서 나중에 채우는 방식보다, 처음부터 `first = -1`, `last = -1`로 두는 게 안전하다.

* **시간복잡도:**
  왼쪽 경계 한 번, 오른쪽 경계 한 번 찾으면 `O(log n) + O(log n)` → `O(log n)`.

* **경계조건:**
  빈 배열 `[]`
  target 없음
  target이 하나만 있음
  target이 맨 앞에 몰려 있음
  target이 맨 뒤에 몰려 있음
  배열 전체가 target임

5. **다음 개선 방향**

* **다음에 비슷한 문제를 만나면 먼저 확인할 것:**
  “target을 찾으면 끝인가?” 아니면 “target이 들어갈 위치 / 처음 위치 / 마지막 위치를 찾아야 하는가?”

* **반복해서 연습할 포인트:**
  `nums[mid] == target`일 때 바로 반환하지 않고,
  `정답 후보 저장 + 탐색 방향 유지`하는 연습.

* **한 줄 복습 메모:**
  `first/last position은 값을 찾는 문제가 아니라, target 구간의 양쪽 벽을 찾는 문제다.`

```python
class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        left = 0
        right = len(nums)-1
        result = [] # 시작, 마지막 index 저장
        while left<=right:
            mid = (left+right) //2
    
            if nums[mid] == target:
                # 일단 찾음
                # target이 여러개라면 옆에 존재 => 정렬되었으니
                # target이 존재하는 index의 최솟값과 최댓값
                #일단 left ~ right 범위 안에 있긴할거임
            return result

            if nums[mid] < target:
                left = mid + 1

            else:
                right = mid - 1
        
        return [-1,-1]
```
