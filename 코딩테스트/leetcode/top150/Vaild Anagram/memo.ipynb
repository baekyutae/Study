문자열 s와 t 가 주어졌을때 t가 s의 anagram 이면
true를 아니면 false를 반환

anagaram: 문자는 그대로 사용하되 순서를 재배열하여 만든 새로운 글자

true이려면 각문자를 사용한 횟수가 같으면 됨

```
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        문자열 s에서 한글자씩 받아와
            글자별 사용횟수를 기록

        문자열 t도 한글자씩 받아와
            글자별 사용횟수를 기록

        두저장소가 같으면 true
        아니면 false

```

````python
a = {1:2,2:3}
b = {2:3,1:2}

print(a==b)
````
