유형: 해시(Hash), 더 구체적으로는 해시 세트(Hash Set)

막힌 이유: 시간복잡도 해결을 위한 자료구조 전환 실패. 정렬을 하면 $O(N \log N)$이 되기 때문에 안 된다는 것은 매우 정확하게 파악하셨습니다. 하지만 '정렬을 하지 않고 어떻게 크기 순서를 알 수 있지?'라는 생각에 갇혀, 주어진 배열(list) 자체에서만 어떻게든 답을 찾으려 했던 것이 더 이상 진행이 안 된 원인입니다. 순서대로 나열하는 것 대신, 다음 숫자가 존재하는지만 빠르게 확인하면 연속성을 파악할 수 있다는 발상의 전환이 필요했습니다.

트리거: **"정렬되지 않은 데이터에서 특정 값의 탐색을 $O(N)$ 안에 끝내야 한다"**는 조건이 주어지면, 가장 먼저 **Hash(Set 또는 Dict)**를 떠올리셔야 합니다. 특히 이 문제처럼 나보다 1 작은 수, 나보다 1 큰 수가 있는지 없는지 그 존재 여부를 묻는 상황은 Set을 꺼내 들어야 하는 아주 강력한 신호입니다.

파이썬 포인트: set()과 in 연산자. 파이썬에서 리스트(List)를 대상으로 x in lst를 수행하면 최악의 경우 리스트 끝까지 찾아야 하므로 $O(N)$의 시간이 걸립니다. 하지만 배열을 num_set = set(nums)로 변환한 뒤 x in num_set을 수행하면, 파이썬 내부의 해시 테이블 구조 덕분에 평균적으로 **$O(1)$**만에 찾아낼 수 있습니다. 이 탐색 시간의 차이를 아는 것이 파이썬 알고리즘 풀이의 핵심 포인트입니다.

주어진 숫자 배열 nums의 원소들로 만들수 있는 연속적인 숫자 배치의 최대 길이를 반환
