````python
네가 이 문제에서 막힌 핵심은, **정렬되어 있음**을 “검색 범위를 단조롭게 줄일 수 있는 신호”로 못 연결한 거야.

먼저 네가 물은 브루트포스부터 보면:

* 모든 쌍을 다 보면 시간복잡도는 **O(n²)**
* 원소가 최대 `10^4`개면 비교 쌍 수는 대략 `10^4 * 10^4 = 10^8`급, 정확히는 절반 수준이라도 **약 5천만 쌍**
* 그래서 이 문제 의도 풀이로는 비효율적이야

이제 네 풀이가 왜 막혔는지 기준에 맞춰 정리해볼게.

---

## 1. 유형

**투포인터**

이 문제는 해시맵도 못 쓰고, 배열이 **오름차순 정렬**되어 있으니까
양 끝에서 시작해서 합을 비교하며 범위를 줄이는 **투포인터**가 정석이야.

왜냐면 정렬되어 있으면:

* 합이 너무 작다 → 더 큰 수가 필요함 → **왼쪽 포인터를 오른쪽으로 이동**
* 합이 너무 크다 → 더 작은 수가 필요함 → **오른쪽 포인터를 왼쪽으로 이동**

이게 항상 성립해.

---

## 2. 막힌 이유

### A. **시간복잡도 감각이 아직 불안정함**

네가 “브루트 포스로 하면 얼마나 나올려나?”에서 멈춘 건,
아직 **쌍 비교 = O(n²)**가 바로 안 떠오른 상태라는 뜻이야.

이 문제는 조건에 이미 힌트가 있었어:

* 추가 공간 O(1)
* 정렬되어 있음
* 정답 하나 보장

이 조합이면 “브루트포스나 해시맵 말고 다른 구조를 써야 한다”로 가야 했어.

---

### B. **정렬 정보를 ‘가지치기’ 수준으로만 보고, ‘이동 규칙’으로 못 바꿈**

네 사고 흐름은 이런 느낌이었어:

* `target`보다 큰 것들 좀 제외
* 왼쪽 숫자들에서 찾아본다
* 불필요한 숫자들을 처낸 다음 생각한다

이건 나쁜 접근은 아닌데,
문제는 **무엇을 기준으로 버릴지 규칙이 명확하지 않다**는 거야.

투포인터는 그냥 “좀 처내자”가 아니라:

* 현재 합 `numbers[left] + numbers[right]`를 본다
* 작으면 left++
* 크면 right--

처럼 **항상 다음 행동이 결정되는 규칙**이 있어.

네 접근은 “부분적으로 줄여보자”까지는 갔는데,
그다음 **어느 쪽을 움직여야 하는지 근거 있는 규칙**이 없어서 막힌 거야.

---

### C. **원소 하나의 크기로 판단하려고 함**

예를 들어 네가 한 생각:

> target이 9니까 9보다 큰 것들 제외하고...

이건 pair sum 문제에서 위험한 사고야.
중요한 건 **개별 원소의 크기**가 아니라 **두 수의 합**이거든.

그리고 더 중요한 건,
이 문제는 **정렬을 이용해서 “합이 크면 오른쪽 줄이기 / 작으면 왼쪽 키우기”**를 해야지,
“큰 수를 미리 싹 제거”하는 방식으로 풀리는 문제가 아니야.

즉, 네가 문제를 **값 필터링 문제**처럼 봤고,
실제로는 **정렬된 배열에서의 방향성 탐색 문제**였어.

---

### D. **문제 조건을 완전히 활용하지 못함**

이 문제의 핵심 조건은 세 개야.

1. **정렬되어 있다**
2. **추가 공간 O(1)**
3. **정답이 정확히 하나다**

이 세 개를 합치면 거의 바로 투포인터여야 해.

특히 `O(1)` 추가 공간 제한은
“해시맵 쓰지 마라”는 뜻이고,
그럼 남는 대표 패턴이 **투포인터**야.

즉, 막힌 이유를 한 줄로 말하면:

> **문제 조건을 각각 따로 읽었지, 하나로 묶어서 패턴으로 인식하지 못했다.**

---

## 3. 트리거

다음엔 이런 신호가 보이면 **투포인터**를 바로 의심하면 돼.

### 꼭 떠올려야 할 신호

* **배열이 정렬되어 있음**
* **두 수의 합 / 차 / 조건 비교**
* **추가 공간 제한 O(1)**
* **연속 부분배열이 아니라, 정렬된 배열의 양끝 비교가 가능함**

이 문제에서는 거의 이렇게 읽어야 해:

> “정렬된 배열 + 두 수의 합 + 해시맵 금지”
> → **투포인터 써라**

---

## 4. 파이썬 포인트

이 문제에서 중요한 파이썬 포인트는 많지 않아.

### 핵심

* `left`, `right` 포인터 2개
* `while left < right`
* 합 비교 후 포인터 이동
* 인덱스는 **1부터 반환**해야 하므로 `left + 1`, `right + 1`

예시 형태는 이런 흐름이야:

```python
left, right = 0, len(numbers) - 1

while left < right:
    s = numbers[left] + numbers[right]

    if s == target:
        return [left + 1, right + 1]
    elif s < target:
        left += 1
    else:
        right -= 1
```

### 여기서 파이썬적으로 주의할 것

* `enumerate`, `set`, `dict`, `defaultdict` 같은 건 오히려 이 문제에선 안 맞음
* 이유: 추가 공간 O(1) 조건 때문

---

## 5. 네가 왜 못 풀었는지 한 문장 요약

> **정렬을 이용한 “단조로운 포인터 이동 규칙”을 못 잡고, 값 몇 개를 미리 제외하는 식의 부분 가지치기로 접근해서 막혔다.**

---

## 6. 오답노트용으로 짧게 정리

### 유형

* **투포인터**

### 막힌 이유

* **시간복잡도 계산이 불안정했음**
* **정렬의 의미를 “양끝에서 범위를 줄일 수 있음”으로 연결하지 못했음**
* **개별 원소 필터링으로 접근해서, 포인터 이동 규칙을 못 세움**

### 트리거

* **정렬된 배열**
* **두 수의 합**
* **추가 공간 O(1) 제한**
* 이 조합이면 **투포인터** 먼저 의심

### 파이썬 포인트

* `left`, `right` 두 포인터
* `while left < right`
* `s < target`이면 `left += 1`
* `s > target`이면 `right -= 1`
* 반환 시 `1-based index` 처리: `[left+1, right+1]`

---

## 7. 네 사고에서 좋았던 점도 있음

완전히 엉뚱하게 간 건 아니야.

* “불필요한 숫자를 처내야 하나?”라고 생각한 건 좋았어
* 즉, **브루트포스가 아니라 범위를 줄여야 한다**는 감각은 있었음

다만 다음 단계가 틀렸던 거야.

* 네 생각: “어떤 숫자들을 제거할까?”
* 정답 방향: “현재 합을 보고 어느 포인터를 움직일까?”

이 차이만 잡으면 비슷한 문제는 훨씬 잘 풀릴 거야.

원하면 내가 다음 답변에서 이 문제를 기준으로
**왜 `sum < target`이면 left를 움직이고, `sum > target`이면 right를 움직여도 절대 정답을 놓치지 않는지**까지 그림처럼 설명해줄게.

````
