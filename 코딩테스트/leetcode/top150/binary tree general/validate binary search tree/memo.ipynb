1. **한 줄 진단**

* BST를 “부모-자식 1단계 비교” 문제로 봐서, **조상 노드가 만든 허용 범위**를 상태로 들고 내려가는 발상을 못 잡았다. LeetCode의 BST 조건은 단순히 직계 자식만이 아니라 “왼쪽/오른쪽 **서브트리 전체**”가 기준값보다 작거나 커야 한다는 조건이다. ([LeetCode][1])

---

2. **내 사고가 막힌 지점**

* **처음 접근:**
  재귀로 왼쪽 서브트리, 오른쪽 서브트리를 각각 검사하려고 함.

* **맞았던 생각:**

  * 왼쪽은 부모보다 작아야 함
  * 오른쪽은 부모보다 커야 함
  * 자식 노드도 같은 규칙을 만족해야 함
  * 재귀로 풀어야 한다는 방향은 맞음

* **틀렸던 생각:**

  * “현재 부모와 바로 아래 자식만 비교하면 된다”고 좁게 봄
  * `좌측`, `우측` 결과만 합치면 된다고 생각함
  * `부모 val을 어떻게 알지?`에서 막혔는데, 진짜 핵심은 부모 하나가 아니라 **상한/하한 범위**였음

* **결정적으로 막힌 이유:**
  BST 검증의 핵심 상태는 `현재 노드가 만족해야 하는 min/max 범위`인데, 이걸 함수 인자로 설계하지 못했다.

---

3. **실패 유형 / 등급**

* **주 실패 유형:**

  * 상태·불변식 설계 실패

* **등급:**

  * D 해설 따라가지만 자력 불가

* **근거:**
  재귀 호출 구조는 떠올렸지만, `현재 노드가 어떤 범위 안에 있어야 하는가`라는 불변식을 못 세웠다. 또한 `우측 = self.TisValidBS(root.left)`처럼 오른쪽 재귀 호출도 왼쪽으로 쓰는 구현 실수 징후가 있다.

---

4. **일반화해서 공부할 요소**

* **알고리즘/자료구조 신호:**
  트리에서 “서브트리 전체가 조건을 만족해야 한다” → 단순 자식 비교가 아니라 **조건을 아래로 전달**해야 함.

* **상태/불변식:**
  각 노드는 `lower < node.val < upper`를 만족해야 한다.

* **파이썬 문법/구현 포인트:**
  재귀 함수에 `node, lower, upper` 같은 추가 상태를 넘기는 연습 필요.

* **시간복잡도:**
  모든 노드를 한 번씩 확인하므로 `O(N)`.

* **경계조건:**

  * `None` 노드는 검증할 값이 없으므로 True 취급
  * 중복값은 허용 안 됨
  * 직계 자식은 맞아도 손자/후손이 조상 조건을 깨면 False

5. **다음 개선 방향**

* **다음에 비슷한 문제를 만나면 먼저 확인할 것:**
  “이 조건이 바로 아래 자식만 보면 되는가, 아니면 조상/경로 전체의 제약을 받아야 하는가?”

* **반복해서 연습할 포인트:**
  재귀 함수에 단순히 `node`만 넘기지 말고, `현재 노드가 지켜야 할 조건`도 같이 넘기는 연습.

* **한 줄 복습 메모:**
  BST 검증은 부모-자식 비교가 아니라, **조상이 만든 min/max 범위를 모든 노드가 지키는지 확인하는 문제**다.
