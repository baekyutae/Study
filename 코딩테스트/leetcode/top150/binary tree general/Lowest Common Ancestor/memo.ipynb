2. **내 사고가 막힌 지점**

* **처음 접근:**
  `p, q를 자식으로 가지는 노드`를 찾고, 그중 `root에서 가장 먼 노드`를 반환하려 함.

* **맞았던 생각:**

  * p, q가 트리에 무조건 존재한다는 조건 확인함.
  * 자기 자신이 정답이 될 수 있다는 점 인식함.
  * 어떤 노드 기준으로 아래에 p, q가 모두 있는지 확인해야 한다는 방향은 맞음.

* **틀렸던 생각:**

  * “자식으로 가진다”를 직접 체크하려 함.
    실제로는 **왼쪽 서브트리에서 p/q 발견 여부**, **오른쪽 서브트리에서 p/q 발견 여부**가 중요함.
  * depth 카운트를 따로 두고 “가장 깊은 후보”를 고르려 함.
    이 문제는 깊이를 비교하는 문제가 아니라, **재귀가 아래에서 위로 반환되면서 처음 만나는 분기점**을 찾는 문제임.

* **결정적으로 막힌 이유:**
  `노드가 무엇을 반환해야 하는지`를 못 정했다.
  즉, 함수의 반환값을 “정답 노드 or 발견한 p/q or None”으로 설계하지 못해서 구현으로 못 들어갔다.

4. **일반화해서 공부할 요소**

* **알고리즘/자료구조 신호:**

  * 이진트리에서 “두 노드의 공통 조상”
  * “왼쪽 결과 + 오른쪽 결과를 합쳐 현재 노드 판단”
  * 후위 순회형 재귀

* **상태/불변식:**

  * 현재 함수는 `현재 root 아래에서 p 또는 q 또는 LCA를 찾으면 그 노드를 반환한다`.
  * 왼쪽과 오른쪽에서 각각 값이 나오면 현재 root가 LCA.
  * 한쪽만 나오면 그 값을 위로 전달.
  * 둘 다 없으면 None.

* **파이썬 문법/구현 포인트:**

  * `if root is None`
  * `if root == p or root == q`
  * `left = self.lowestCommonAncestor(root.left, p, q)`
  * `right = self.lowestCommonAncestor(root.right, p, q)`

* **시간복잡도:**

  * 모든 노드를 최대 한 번 방문하므로 O(N)

* **경계조건:**

  * root 자체가 p 또는 q인 경우
  * p, q가 서로 다른 서브트리에 있는 경우
  * p가 q의 조상인 경우
  * 한쪽 서브트리에 둘 다 있는 경우
