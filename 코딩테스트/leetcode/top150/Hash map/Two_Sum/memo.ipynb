좋아. 네가 이번 문제를 못 푼 이유를 딱 저 4가지 기준으로 정리하면 이렇게 볼 수 있어.

---

## 유형

**해시맵(Hash Map)**
보조적으로는 **정렬 + 투포인터**도 가능하지만,
이 문제의 정석 유형은 **해시맵으로 보완값(complement) 찾기**야.

네가 처음에 떠올린 **투포인터**는
`정렬된 배열` 문제에서 자주 쓰는 패턴이고,
이번 문제는 **정렬되지 않은 배열의 원래 인덱스**를 반환해야 해서
투포인터를 바로 적용하기엔 유형 선택이 어긋났어.

---

## 막힌 이유

핵심 원인은 **풀이 유형 오인 + 구현 실수** 쪽에 가까워.

### 1. 정렬되지 않은 배열에 투포인터를 적용하려고 함

투포인터는 보통 이런 전제가 있어:

* 배열이 정렬되어 있음
* 왼쪽을 키우면 합이 커지고
* 오른쪽을 줄이면 합이 작아짐

근데 지금 문제는 비정렬 배열이라
포인터를 옮겼을 때 합이 어떤 방향으로 변할지 보장이 없음.

즉,
**“합이 크면 큰 값을 줄인다 / 합이 작으면 작은 값을 늘린다”**
이 논리가 성립하지 않는 구조였어.

---

### 2. 문제의 진짜 요구사항을 끝까지 못 붙잡음

이 문제는 단순히 두 수를 찾는 게 아니라:

* **원래 배열 기준 인덱스 반환**
* **정답 조합이 반드시 존재**
* **비정렬 배열**

이 3개가 중요해.

그런데 네 사고 흐름은
“두 수의 합을 맞춘다”에 집중해서
“원래 인덱스를 어떻게 안정적으로 보존할까?”가 뒤로 밀렸어.

---

### 3. 정렬 후 처리할 때 기준이 섞임

두 번째 시도에서:

* 원본 배열 `nums`
* 정렬 배열 `s_nums`
* 값→인덱스 맵 `map`

이 3개를 동시에 쓰면서
포인터 기준이 섞였어.

즉,

* 포인터는 `s_nums` 기준인데
* 비교는 `nums[left]`, `nums[right]`
* 반환도 `nums[left]`, `nums[right]`

이렇게 되니까 논리 축이 무너진 거야.

이건 전형적인 **구현 실수**야.

---

### 4. 중복값 처리에 대한 감각이 아직 약했음

`map[i] = n` 방식은 값이 중복되면 덮어써져.

예:

```python
nums = [3, 3]
```

이 경우 값 3 하나에 인덱스 2개가 대응되는데
딕셔너리 하나로는 마지막 것만 남게 돼.

즉,
**값 하나당 인덱스 하나라고 가정한 점**도 막힌 이유 중 하나야.

---

## 트리거

다음 신호가 보이면 **해시맵**을 먼저 떠올려야 해.

### 1. “두 수의 합이 target”

이 문구가 나오면 거의 자동으로
**“현재 값 말고, 나머지 하나는 target - num 이네?”**
를 떠올려야 해.

즉,

* 지금 값이 `num`
* 필요한 값은 `target - num`

이 보완값 사고가 바로 트리거야.

---

### 2. “원래 인덱스를 반환”

정렬해버리면 원래 위치 정보가 흐트러짐.
그래서 이 문구가 보이면
정렬보다 **해시맵으로 원래 인덱스를 바로 저장**하는 쪽이 자연스러워.

---

### 3. “정답이 반드시 하나 존재”

이 조건이 있으면 더 편해져.

왜냐면:

* 여러 경우 비교할 필요 적고
* 찾자마자 바로 반환 가능
* 해시맵 1패스가 매우 잘 맞음

---

### 4. 배열이 비정렬인데 투포인터가 떠오를 때

이때 스스로 바로 체크해야 해:

> “이 배열은 정렬되어 있나?”
> “포인터 이동이 합의 증가/감소를 보장하나?”

둘 중 하나라도 아니면
투포인터는 보류해야 해.

**투포인터를 떠올린 것 자체보다,

## 오답노트용으로 짧게 쓰면

* **유형:** 해시맵
* **막힌 이유:** 비정렬 배열에 투포인터를 적용하려 했고, 정렬 후 원본 인덱스 보존/중복값 처리에서 구현이 꼬임
* **트리거:** “두 수의 합 = target”, “원래 인덱스 반환”, “정답 1개 보장” → 보완값 해시맵 떠올리기
* **파이썬 포인트:** `enumerate`, `dict`, `target - num`, `(값, 인덱스)` 튜플 정렬

원하면 내가 다음 답변에서 이걸 **네 오답노트 스타일로 바로 붙여넣을 수 있게 더 짧고 예쁘게 정리**해줄게.

From training data.


배열이 주어졌을때 합이 target이 되는 두 원소의 index를 리스트 형태로 출력
정답인 조합이 무족건 존재


양끝에 두좌표를 둠

두좌표에 원소를 더함

target 보다 크다?

두원소중 더 큰값을 안쪽으로 이동

target 보다 작다?

두 원소중 더 작은 값을 안쪽으로 이동

위 과정을 left < right인동안 반복해 좌표를 찾는다.

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        left = 0
        right = len(nums)-1

        while left< right:
            current_sum = nums[left]+nums[right]
            if current_sum == target:
                return [left,right]
            # target 보다 큰경우
            if current_sum > target:
                if nums[left] > nums[right]:
                    left+=1
                else:
                    right -= 1
            # target 보다 작은경우
            else:
                if nums[left] > nums[right]:
                    right -= 1
                else:
                    left+=1
```
이렇게 하니 안되는게 정렬된 배열이 아니라서 

[3,2,4] 입력 들어왔을 때 왼쪽이 이동해야하는데 안쪽이 이동해서 null 반환


enum으로 번호를 붙여

0:3, 1:2, 2:4

정렬을 해

2 3 4 타겟이 6

2, 5 , 7 , 12, 13 타겟이 19

이러면 원래 논리 대로 좌우 좁혀가며 찾는게 가능할듯?...
좌표는 enum 붙였던걸로 반환

```python

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        left = 0
        right = len(nums)-1
        map = {}
        for n,i in enumerate(nums):
            map[i] = n

        s_nums = sorted(nums)

        while left<right:
            c_sum = s_nums[left]+s_nums[right]
            if c_sum == target:
                return [map[nums[left]], map[nums[right]] ]

            if c_sum > target:
                if nums[left] > nums[right]:
                    left+=1
                else:
                    right -= 1
            # target 보다 작거나 같은 경우
            else:
                if nums[left] > nums[right]:
                    right -= 1
                else:
                    left+=1


```
