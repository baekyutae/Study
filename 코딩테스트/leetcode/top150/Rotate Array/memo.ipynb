nums에서 0부터 끝에서 한칸 앞 원소까지 index를 1증가 시켜 덮어 씌움  => 이게 문제가 앞에서 부터 덮어 씌우면 
[1,2,3,4] -> [1,1,3,4] -> [1,1,1,4]->[1,1,1,1] 이렇게 되버림

뒤에서 부터 덮어 씌워야함
4는 따로 저장해두고
[1,2,3,4] -> [1,2,3,3]->[1,2,2,3]->[1,1,2,3,] -> [4,1,2,3]

단 0부터 k만큼 반복해야 하니깐 k 값이 크면 클수록 비효율 발생 O(kn)

대안으로 뒤집는 함수를 만들어
전체를 뒤집고
앞에서부터 k개 먼저 뒤집고
남은 원소들을 뒤집음
k = 3
nums = [1,2,3,4,5,6,7]

-> [7,6,5,4,3,2,1]
-> [(5,6,7),4,3,2,1]
-> [5,6,7,(1,2,3,4)]

결과 : [5,6,7,1,2,3,4]

````python
from typing import List

class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        n = len(nums)
        k %= n  # k가 n보다 클 수 있으니까

        for _ in range(k):
            last = nums[-1]  # 맨 뒤 값 저장

            # 뒤에서부터 한 칸씩 오른쪽으로 밀기
            for i in range(n - 1, 0, -1):
                nums[i] = nums[i - 1]

            nums[0] = last

answer = Solution()
nums = [1,2,3,4,5,6,7]
k = 3
print(answer.rotate(nums, k))
````

````python
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        n = len(nums)
        k %= n

        def reverse(left: int, right: int) -> None:
            while left < right:
                nums[left], nums[right] = nums[right], nums[left]
                left += 1
                right -= 1

        reverse(0, n - 1)
        reverse(0, k - 1)
        reverse(k, n - 1)
````
