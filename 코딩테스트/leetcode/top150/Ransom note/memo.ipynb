ransomnote와 magazine이란 두 문자열이 주어짐

magazine에 문자를 조합해 ransomnote와 동일한 문자열을 만들수 있으면 true 안되면 false 반환

중복 사용은 안됨

ransomnote의 문자 구성을 먼저 파악해야함

aab라면

a 두번 b 한번 사용

즉 ransomnote에 a 두개 b 하나가 있으면 만들 수 있음

```plaintext
def canConstruct(self, ransomNote: str, magazine: str) -> bool:
    ransomnote의 구성을 딕셔너리로 만들어 파악

    magazine도 알파벳: 사용 횟수 딕셔너리로 만듬

    magazine에 문자 구성이 ransomnote를 만들기에 충분한지 체크

    =>   ransomnote 딕셔너리에서 문자 하나를 꺼냄
            key값이 magazine에 존재하는지
                존재한다면 value값이 magazine >= ransomnote 인지

                하나의 경우라도 존재 x 또는 일치 하지 않으면 false 반환

        끝까지 확인 했다면 true

```
aabccc , aabbbccc


````python
class Solution:
    def canConstruct(self, ransomNote: str, magazine: str) -> bool:
        r = {}
        for i in ransomNote:
            if i in r:
                r[i] += 1
            else:
                r[i] = 1

        m = {}
        for i in magazine:
            if i in m:
                m[i] += 1
            else:
                m[i] = 1

        for check in r:
            if check in m:
                if m[check]>=r[check]:
                    continue
                else:
                    return False
            else:
                return False
        return True
        
````
