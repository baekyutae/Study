

## 유형

이 문제의 핵심 유형은

**배열 + prefix/suffix(누적값 재사용)** 이다.

조금 더 정확히 말하면:

* BFS / DP / 그리디 / 투포인터 쪽이라기보다는
* **누적곱(prefix product, suffix product)** 을 이용하는 배열 문제
* “각 칸의 답을 독립적으로 다시 계산하지 말고, 왼쪽 정보와 오른쪽 정보를 재사용”하는 문제

즉, 이 문제를 못 푼 건
투포인터나 DP를 몰라서라기보다
**누적 정보 설계형 배열 문제**에 아직 익숙하지 않아서라고 보는 게 맞다.

---

## 막힌 이유

네가 막힌 핵심 이유는 **상태정의를 못 한 것**에 가깝다.

정확히는 이런 흐름이었음:

* 브루트포스는 바로 떠올림
* `O(n^2)`라는 것도 감지함
* 그런데 그 다음 단계인
  **“각 칸의 답을 어떤 중간 상태로 표현할 수 있지?”**
  여기서 멈춤

이 문제의 정답 상태는 사실

`answer[i] = i의 왼쪽 곱 * i의 오른쪽 곱`

인데,
이걸 상태로 못 잡으면 계속

* “하나 빼고 나머지 다 곱해야 하나?”
* “그걸 더 빠르게 어케 하지?”

여기서 맴돌게 됨.

즉 정리하면:

* **시간복잡도 계산 실패**는 아님
  → 오히려 `O(n^2)` 위험을 잘 감지했음
* **구현 실수**도 아님
  → 구현까지 못 간 상태
* 진짜 원인은
  → **정답을 만들 중간 상태(왼쪽 곱 / 오른쪽 곱)를 정의하지 못함**

## 트리거

앞으로 이런 문제에서 떠올려야 하는 신호는 이거야:

### 1. 각 위치의 답을 매번 새로 계산하면 너무 비쌀 때

이 문제는 각 칸마다 “나 제외 전부 곱하기”를 하면 `O(n^2)`가 됨.
이때 바로

**“중복 계산을 재사용해야 한다”**
를 떠올려야 함.

---

### 2. 각 위치의 답이 왼쪽 부분과 오른쪽 부분으로 분해될 때

이 문제가 특별한 이유는
`i`를 제외한 전체가 자연스럽게

* `i`의 왼쪽 부분
* `i`의 오른쪽 부분

으로 나뉜다는 점이야.

즉 이런 형태가 보이면 신호임:

* 나 제외 나머지
* 현재 위치 기준 왼쪽/오른쪽
* 구간 정보 결합
* 각 칸마다 주변 정보가 필요함

이런 문구가 보이면
**prefix / suffix**를 의심해야 함.

---

### 3. 나눗셈 없이 전체 효과를 구해야 할 때

많이들 처음엔
“전체 곱 구하고 나로 나누면 되잖아?”
를 떠올리는데, 문제에서 보통 나눗셈 제한이 있거나 0 때문에 막힘.

이럴 때 떠올릴 패턴:

**전체값 1개로 처리 못 하면, 좌측 누적 + 우측 누적으로 분해하자**

## 파이썬 포인트

이 문제에서 중요한 건 어려운 문법이 아니라
**배열과 반복문의 의미를 정확히 다루는 것**이야.

핵심 포인트는:

### `answer = [1] * n`

정답 배열을 미리 1로 채워놓고 시작
곱셈에서는 1이 항등원이라 누적곱 시작값으로 적절함

---

### 정방향 루프

```python
for i in range(n):
```

왼쪽 누적곱을 채우는 용도

---

### 역방향 루프

```python
for i in range(n - 1, -1, -1):
```

오른쪽 누적곱을 채우는 용도

이 역순 순회가 익숙하지 않으면 구현이 잘 안 떠오를 수 있음

---

### 누적 변수 업데이트 순서

```python
answer[i] = left_product
left_product *= nums[i]
```

순서가 중요함.
현재 칸에는 **현재까지의 왼쪽 곱만** 넣고,
그 다음에 현재 값을 곱해서 다음 칸 준비를 해야 함.

오른쪽도 마찬가지:

```python
answer[i] *= right_product
right_product *= nums[i]
```

즉 이 문제의 파이썬 포인트는
`defaultdict`, `set` 같은 자료구조가 아니라

* 배열 초기화
* 정순/역순 순회
* 누적 변수
* 업데이트 순서 제어

이 4개다.
