LRU 캐시 규칙을 따르는 자료구조를 설계

__init__  생성자 에선

일단 key와 value 두개가 클래스 변수로 들어감
그리고 최대 저장 용량인 capacity도 들어감

구현해야할 메서드는

get, put 두 함수를 구현해야함

get: key가 존재하면 value를 반환 없으면 -1을 반환

put: key와 value를 쌍을 지어서 저장,

    핵심은 
    1. capacity 초과 여부를 확인 
    2. 오래된 key-value 식별
    3. 해당 값 제거

get과 put 함수는 각각 평균적으로 O(1) 시간복잡도로 실행

------------------------
class Container:
    def __init_(self,key,value,next):
        self.key = key
        self.value = value
        self.next= None

class LRUCache:

    def __init__(self, capacity: int):
        self.capacity = capacity
        

    def get(self, key: int) -> int:
        만약 key가 들어 있으면:
            링크드 리스트에서 해당 key-value를 tail로 이동
            return value

        else:
            return -1
    
    def put(self, key: int, value: int) -> None:
        if capacity를 초과:
            제일 오래된 key-value 식별
            => 링크드 리스트에서 head를 제거
            
            새 key - value를 tail에 넣음

        else:
            새 key - value 넣음
