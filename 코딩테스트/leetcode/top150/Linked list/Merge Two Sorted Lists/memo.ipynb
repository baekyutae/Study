  # 2. 사고 흐름과 막힌 지점

  내 최초 가설:
  list2의 각 노드를 list1 안에서 알맞은 위치를 찾아 끼워 넣는다. 노드 수가 최대 50이라 비교가 많아도 괜찮다고 봤다.

  막힌 지점:
  “어디에 끼워 넣을지”는 생각했지만, 끼워 넣은 뒤 기존 list1.next, list2.next를 어떻게 보존하고 이동할지 설계되지 않았다. 그래서 list1.next = list2 이후 원
  래 연결이 끊기고, 다음 노드로 진행하는 로직도 완성되지 않았다.

  맞았던 생각:
  정렬되어 있으므로 값 비교로 병합할 수 있다는 점은 맞았다.
  입력 크기가 작아서 비효율적인 방식도 통과 가능하다는 복잡도 감각도 크게 틀리지는 않았다.

  틀렸던 생각:
  삽입 위치 탐색을 중심으로 생각한 것이 구현 난도를 키웠다.
  이 문제의 핵심은 “정렬된 두 흐름 중 작은 노드를 하나씩 결과 뒤에 붙인다”인데, 사고가 “한 리스트 안에 다른 리스트를 삽입한다”로 기울었다.
  또 list1, list2가 None일 수 있는 경계조건이 코드에 반영되지 않았다.

  # 4. 일반화 포인트

  알고리즘/자료구조 신호:
  두 입력이 이미 정렬되어 있고, 하나의 정렬 결과를 만들어야 한다. 이 신호는 “두 포인터로 앞에서부터 작은 값을 선택”하는 패턴을 떠올려야 한다.

  상태/불변식:
  결과 리스트의 마지막 노드 tail을 유지한다.
  tail 뒤에는 아직 확정되지 않은 노드만 남아 있어야 한다.
  매 단계에서 list1.val과 list2.val 중 작은 노드를 tail.next에 붙이고, 붙인 쪽 포인터만 한 칸 이동하면 충분하다.

  복잡도/경계조건:
  각 노드를 한 번씩만 붙이면 되므로 O(n + m)이다.
  한쪽 리스트가 먼저 끝나면 남은 리스트는 이미 정렬되어 있으므로 통째로 붙이면 된다.
  일반화 문장:
  “두 정렬된 흐름을 합칠 때는 삽입 위치를 매번 다시 찾지 말고, 결과의 끝 포인터를 유지하며 작은 쪽을 하나씩 소비한다.”

  - tail 이전은 이미 정렬 완료다.
  - list1, list2는 아직 결과에 붙이지 않은 남은 구간의 시작이다.
  - 둘 중 작은 노드 하나만 붙이면 정렬성이 유지된다.

  변형 대응 훈련:

  노드를 재연결할 때는 “기존 next를 잃지 않는가?”를 확인해라.
  삽입 방식보다 dummy head + tail 방식으로 먼저 풀어보는 훈련이 필요하다.
