# 오답노트: Reverse Linked List II

## 1. 왜 못 풀었는가

### 1) 반환값을 잘못 이해함

문제 화면에는 결과가 `[1,4,3,2,5]`처럼 보이지만, 실제로 반환해야 하는 건 **Python list가 아니라 링크드 리스트의 head 노드**다.

즉, `return result`가 아니라 최종 연결이 끝난 linked list의 head를 반환해야 한다.

---

### 2) 링크드 리스트를 배열처럼 생각함

처음 접근은 노드를 딕셔너리에 저장한 뒤,

```python
position[left], position[right] = position[right], position[left]
```

처럼 위치를 바꾸는 방식이었다.

하지만 링크드 리스트의 순서는 **인덱스가 아니라 `.next` 연결**로 결정된다.

딕셔너리 안에서 노드 위치를 바꿔도 실제 리스트 연결은 그대로다.

```text
배열: index 순서가 실제 순서
링크드 리스트: next 연결이 실제 순서
```

---

### 3) 단방향 리스트에서 right를 왼쪽으로 이동시키려 함

배열이면 `left`, `right`를 양쪽에서 좁히며 swap할 수 있다.

하지만 단방향 링크드 리스트는 이전 노드로 돌아갈 수 없다.

```text
1 -> 2 -> 3 -> 4
```

`4`에서 `3`으로 이동할 방법이 없다.

그래서 이 문제는 **양끝에서 교환하는 문제**가 아니라, **앞에서부터 next 연결을 재배치하는 문제**로 봐야 한다.

---

## 2. 핵심 개념

이 문제의 핵심은:

```text
노드 값을 바꾸는 게 아니라,
next 연결을 바꿔서 left~right 구간만 뒤집는 것
```

이다.

즉, 링크드 리스트 뒤집기는 “값 교환”이 아니라 “화살표 방향 변경” 문제다.

---

## 3. 핵심 로직

예시:

```text
1 -> 2 -> 3 -> 4 -> 5
left = 2, right = 4
```

목표:

```text
1 -> 4 -> 3 -> 2 -> 5
```

이때 핵심은 `2`, `3`, `4`를 양쪽에서 swap하는 게 아니다.

대신:

```text
start 뒤에 있는 노드를 하나씩 떼어서
prev 바로 뒤에 붙인다.
```

초기 상태:

```text
1 -> 2 -> 3 -> 4 -> 5
     ^    ^
   start then

prev = 1
start = 2
then = 3
```

1회차: `3`을 `2` 앞으로 이동

```text
1 -> 3 -> 2 -> 4 -> 5
```

2회차: `4`를 `2` 앞으로 이동

```text
1 -> 4 -> 3 -> 2 -> 5
```

결과:

```text
1 -> 4 -> 3 -> 2 -> 5
```

---

## 4. 꼭 기억할 포인트

`start`는 계속 같은 노드에 머문다.

예시에서는 `start = 2`이고, `2`는 뒤집힌 구간의 맨 뒤로 밀려난다.

```text
처음: 1 -> 2 -> 3 -> 4 -> 5
1회:  1 -> 3 -> 2 -> 4 -> 5
2회:  1 -> 4 -> 3 -> 2 -> 5
```

그래서 `start`를 이동시키는 게 아니라, `start.next`에 있는 노드인 `then`을 계속 앞으로 빼는 구조다.

    left는 다음 노드로 이동
    right는 이전 노드로 한칸이동
        => 이전 노드로는 어떻게 이동하냐
        딕셔너리 형태로 각 노드의 위치: 노드 이런식으로 저장해뒀다 써먹어야 하나



class Solution:
    def reverseBetween(self, head: Optional[ListNode], left: int, right: int) -> Optional[ListNode]:
        cur = head
        position = {}
        index = 1
        while cur is not None:
            position[index] = cur
            cur = cur.next
            index += 1

        while left<right:
            position[left],position[right] = position[right] ,position[left] 
            left += 1
            right -= 1
        
        temp = head
        count = 1
        result = []
        while count < len(position):
            position[count].next = position[count+1]
            count+=1
        
        n=1
        for _ in range(len(position)+1):
            result.append(position[n].val)
            n+=1

        return  result
