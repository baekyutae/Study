

양의 정수 배열 nums
마찬가지로 양의 정수인 target

nums의 원소를 가지고 배열을 만드는데 원소의 합이 target과 같거나 더 길면서 가장 짧은 배열의 길이를 출력
그런 배열이 안만들어지면 0을 반환

```plain text

예시 1:

입력: target = 7, nums = [2,3,1,2,4,3]
 출력: 2
 설명: 부분 배열 [4,3]은 문제 제약 조건에 따라 길이가 가장 짧습니다.

```

모든 경우의수를 다 찾아서 구한다면 시간복잡도가 얼마나 나올까?

최대 길이가 10만이니 각각 서로다른 10만개의 원소의 합조합 10만^10만 => timeout 나올듯

다른 방법이 필요함

일단 target이 만들어지는지 부터 판별하면 좋을듯

그다음은, min(target == sum(subarray)) 차피 가장 짧은거면 target보다 큰값들은 다 제외

해서 길이를 반환 하면 됨

요지는 부르트 포스를  안하고 배열의 합 조합을 뒤져봐야 하는건데

트리 형태 표를 만들면 어떰

1층 원소 하나만 가지고 만든 배열의 합
2층 원소 두개 가지고 만든 배열의 합 모음
이렇게 하면 근데 결국 부르트 포스네 ...

아 그러면 일케 해보자

스코프를 만드는거임 그래서 스코프의 범위 안에 들어온 원소들을 다 더해보는겨
2칸짜리(점점 늘어남) 스코프를 맨앞에서부터 밀면서 범위 안에 들어온 애들을 더하는거지, 문제는 그러면 한칸 이상 건너뛴 원소들 끼리는 어떻게 더하지?

````python
class Solution:
    def minSubArrayLen(self, target: int, nums: List[int]) -> int:
        left = 0
        current_sum = 0
        ans = float('inf')

        for right in range(len(nums)):
            current_sum += nums[right]

            while current_sum >= target:
                ans = min(ans, right - left + 1) # 윈도우 가장 짧은 길이 갱신
                current_sum -= nums[left] # 합이 target 이상인 동안 left를 계속 줄여서, 조건을 만족하는 가장 짧은 구간을 찾는다
                left += 1 # 윈도우 시작점 한칸 갱신

        return 0 if ans == float('inf') else ans
````
