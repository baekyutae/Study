**유형:** 투포인터 (Two Pointers) / 그리디 (Greedy)

**막힌 이유:** 로직 분리 실패로 인한 구현 실수 (무한 루프 발생)
* **상태 갱신과 포인터 이동의 결합:** 처음에는 "넓이가 기존 최대값보다 커질 때만" 포인터를 이동시키려고 했습니다. 하지만 넓이가 갱신되지 않는 구간(즉, 안쪽 막대기가 더 짧은 구간)을 만나면 포인터가 둘 다 멈춰버리는 논리적 허점이 있었습니다.
* **그리디적 사고의 부재:** 가로 길이가 무조건 1씩 줄어드는 상황에서, 더 큰 넓이를 구하려면 두 막대기 중 **"더 짧은 막대기를 포기(이동)해야만 더 긴 막대기를 만날 가능성이 생긴다"**는 매 스텝마다의 최적의 선택(그리디)을 놓쳤기 때문에 조건 설정에서 헤맸습니다.

**트리거:** 어떤 신호에서 이 풀이를 떠올려야 하는지
* **배열의 크기:** `height` 배열의 길이가 최대 100,000입니다. 이중 `for`문으로 모든 쌍을 확인하는 O(N^2) 완전 탐색을 하면 무조건 시간 초과가 발생하므로, O(N) 또는 O(N log N) 풀이를 떠올려야 합니다.
* **"두 개의 기준점"과 "최대화":** 두 개의 막대기(인덱스)를 선택해야 하고, 가로 길이는 양 끝에서 시작할 때가 가장 깁니다. 가로 길이가 점점 줄어드는 것이 확정된 상태에서 넓이를 최대화하려면, 양 끝에서부터 조건에 따라 좁혀 들어오는 **투포인터** 접근이 가장 직관적입니다.

**파이썬 포인트:** `max()`, `min()` 내장 함수
* 물의 양을 계산할 때 `min(height[left], height[right])`를 사용하고, 최대값을 갱신할 때 `current_max = max(water, current_max)`를 활용한 것은 파이썬의 장점을 아주 잘 살린 부분입니다.
* 지저분한 `if-else` 대소 비교 로직을 내장 함수로 묶어버림으로써 코드가 훨씬 파이썬답고(Pythonic) 가독성이 좋아졌습니다.

양끝에서 시작? ... 거리가 멀수록 좋으니깐

하고 왼쪽에서 한칸 이동해서 물의양 구해보고, 오른쪽에서 한칸 이동해서 물의양 구해보고 비교해서 또 최대값 갱신

 left < right 인동안 이렇게 반복해서 찾아보자


 ```python

class Solution:
    def maxArea(self, height: List[int]) -> int:
        left = 0
        right = len(height)-1
        current_max =  (right - left)*min(height[left],height[right]) 
        while left < right:
            
            

            if ((right-1) - left)*min(height[left],height[right-1]) > current_max:
                right -= 1
                current_max = ((right-1) - left)*min(height[left],height[right])

            if (right - (left+1))*min(height[left+1],height[right]) > current_max:
                left += 1
                current_max = (right - (left+1))*min(height[left+1],height[right])
        
        return current_max
 ```
