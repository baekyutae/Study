push, pop, top, getMin이 가능한 stack class를 만들어야함

pop, top, getMin은 항상 stack이 비어있지 않을 때만 호출됨


```
class MinStack:

    def __init__(self):
        self.stack = []

    def push(self, val: int) -> None:
        self.val = val
        self.stack.append(val)

    def pop(self) -> None:
        self.stack.pop()

    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        return min(self.stack)
        
```

class MinStack:

    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, val: int) -> None:
        
        self.stack.append(val)
        if len(self.stack) == 1:
            self.min_number = val
            self.min_stack.append(self.min_number)

        else:
            if val <= self.min_number:
                self.min_number = val
                self.min_stack.append(self.min_number)


    def pop(self) -> None:
        self.temp = self.stack[-1]
        self.stack.pop()
        if self.temp == self.min_stack[-1]:
            self.min_stack.pop()

    def top(self) -> int:
        return self.stack[-1]
