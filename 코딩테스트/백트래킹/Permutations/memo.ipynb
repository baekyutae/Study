문제:
Permutations

패턴:
Backtracking / DFS

트리거:
“주어진 원소들로 만들 수 있는 **모든 순서의 경우**를 반환하라” → 순열 문제 → 백트래킹을 떠올린다.

핵심 발상:
각 자리마다 아직 사용하지 않은 숫자를 하나 선택하고, 더 깊게 탐색한 뒤, 선택을 되돌리며 다른 경우를 시도한다.

시작점:
아무 숫자도 선택하지 않은 빈 상태에서 시작하거나, swap 방식에서는 `0번 자리`부터 채운다고 생각한다.

자료구조:

* `result`: 완성된 순열들을 저장
* `temp` 방식: 현재까지 선택한 숫자 목록 저장
* `swap` 방식: 별도 목록 없이 원본 배열 자체를 상태로 사용
* `set/visited` 또는 `temp 포함 여부`: 이미 사용한 숫자를 다시 쓰지 않기 위해 사용

방문/상태 처리:

* `temp` 방식: 선택한 숫자를 추가하고, 재귀가 끝나면 제거한다.
* `swap` 방식: 현재 자리에 올 숫자를 swap으로 고정하고, 재귀가 끝나면 다시 swap해서 원복한다.
* 정답 저장 시에는 현재 상태가 이후에 바뀌므로 반드시 복사본을 저장해야 한다.

시간복잡도:
O(n × n!)
순열의 개수가 `n!`개이고, 각 순열을 결과에 저장할 때 길이 `n`만큼 복사하기 때문이다.

반드시 외울 것:

* “모든 순서의 경우” → 순열 → 백트래킹
* 백트래킹은 `선택 → 재귀 탐색 → 선택 취소` 구조다.
* 종료 조건은 “순열 하나가 완성되었을 때”, 즉 선택한 개수가 전체 원소 개수와 같을 때다.
* 이미 사용한 숫자는 다시 선택하면 안 된다.
* swap 방식에서는 `i 이전 구간 = 이미 고정된 구간`, `i 이후 구간 = 아직 선택 가능한 구간`으로 본다.
* 정답에 넣을 때는 현재 리스트의 복사본을 넣어야 한다.
* 시간복잡도는 모든 순열을 실제로 만들어야 하므로 최소 `n!`급이다.

외울 필요 없는 것:

* 정답 코드 전체
* 변수명
* 예제 입력 모양
* 해설자의 표현 그대로
* `temp` 방식과 `swap` 방식 중 특정 구현 하나만 고집하는 것
* 재귀 호출 스택의 모든 세부 순서

내 사고 교정:

* 내가 처음 떠올린 접근:
  “모든 경우의 수니까 백트래킹을 써야 한다”까지는 맞게 떠올렸다.

* 맞았던 생각:
  결과 리스트와 현재 경우를 담는 임시 상태가 필요하다는 생각은 맞았다.
  하나를 선택하고 재귀적으로 다음 선택을 이어간다는 방향도 맞았다.

* 틀렸던 생각:
  종료 조건을 “모든 경우의 수를 result에 담으면”처럼 전체 완료 기준으로 생각했다.
  실제 종료 조건은 전체 탐색 완료가 아니라, **순열 하나가 완성된 순간**이다.
  또한 “첫 번째 숫자 고정”이라는 표현 때문에 for문이 매번 첫 자리만 고르는 것처럼 헷갈렸다. 실제로는 재귀 깊이에 따라 현재 채울 자리가 바뀐다.

* 다음에는 이렇게 생각하기:
  “순열은 자리 채우기 문제다. 현재 자리에서 아직 안 쓴 숫자 하나를 고르고, 다음 자리로 내려간 뒤, 돌아오면 선택을 취소한다.”
