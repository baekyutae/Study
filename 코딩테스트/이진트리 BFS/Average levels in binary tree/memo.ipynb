1. **한 줄 진단**

* BFS의 큰 틀은 잡았지만, **queue에서 꺼낸 현재 노드 `node`를 기준으로 처리해야 한다는 핵심 동작**을 놓쳤다.

---

2. **내 사고가 막힌 지점**

* **처음 접근:**
  `deque`를 만들고, `root`를 넣고, `level_size`로 레벨 단위 평균을 구하려고 함.

* **맞았던 생각:**

  * BFS에는 `queue`를 쓴다.
  * 처음에 `root`를 queue에 넣는다.
  * 레벨마다 `level_size = len(queue)`가 필요하다.
  * 레벨 합계를 구해서 평균을 `average_result`에 넣는다.

* **틀렸던 생각:**

  * 계속 `root.left`, `root.right`를 기준으로 자식을 넣으려 함.
  * `queue.pop()`으로 노드를 꺼냄.
  * `sum += queue.pop()`처럼 노드 객체 자체를 더하려 함.
  * BFS인데 내부 재귀 함수처럼 `def average(...)`를 만들려 함.

* **결정적으로 막힌 이유:**
  BFS의 핵심 단위가 `root`가 아니라, **queue에서 매번 꺼낸 `node`**라는 점을 아직 코드로 연결하지 못했다.

---

3. **실패 유형 / 등급**

* **주 실패 유형:**

  * 개념 이해 부족 + 구현 실패

* **등급:**

  * C 해설 후 백지 구현 가능

* **근거:**
  `queue`, `level_size`, 평균 계산이라는 방향은 맞았음. 다만 BFS의 실제 반복 흐름인 `node = queue.popleft()` → `node.val 처리` → `node.left/right 추가`를 구현하지 못했다.

---

4. **일반화해서 공부할 요소**

* **알고리즘/자료구조 신호:**
  “레벨별”, “층별”, “level order” → BFS + queue.

* **상태/불변식:**
  while문 시작 시점의 queue에는 **현재 레벨의 노드들만 들어 있다**.
  for문 중 추가되는 자식들은 **다음 레벨용**이다.

* **파이썬 문법/구현 포인트:**

  * BFS는 `deque`
  * 꺼낼 때는 `popleft()`
  * 노드 값은 `node.val`
  * `while queue:` 사용
  * `sum`은 내장 함수 이름이라 `level_sum` 같은 이름 권장

* **시간복잡도:**
  모든 노드를 한 번씩 방문하므로 `O(N)`.

* **경계조건:**
  `root is None`이면 빈 리스트 반환을 고려.

---

5. **다음 개선 방향**

* **다음에 비슷한 문제를 만나면 먼저 확인할 것:**
  “지금 처리해야 하는 대상이 `root`인가, queue에서 꺼낸 `node`인가?”

            while queue is not null:

                average = sum/level_size
                average_result.append(average)
