## 1. 한 줄 진단

* `findMedian()` 때마다 전체 데이터를 다시 꺼내 중앙값을 계산하려 해서, **데이터 스트림 문제의 핵심인 “중앙값 상태 유지” 설계를 못 했다.**

---

## 2. 내 사고가 막힌 지점

* **처음 접근:**
  `addNum()`에서 모든 숫자를 heap에 저장하고, `findMedian()`에서 전부 pop해서 정렬된 리스트를 만든 뒤 중앙값 계산.

* **맞았던 생각:**

  * 중앙값은 정렬된 순서 기준으로 구해야 한다는 이해는 맞음.
  * heap에서 계속 pop하면 작은 값부터 나오므로 정렬된 리스트를 만들 수 있다는 생각도 맞음.
  * 짝수 개면 가운데 두 값 평균, 홀수 개면 가운데 값 반환이라는 계산식도 맞음.

* **틀렸던 생각:**

  * `findMedian()` 호출 때마다 전체 데이터를 다시 꺼내도 된다고 생각함.
  * `heappop(self.data)`로 원본 데이터를 직접 비워버림.
  * heap 하나만 있으면 중앙값을 효율적으로 구할 수 있다고 봄.
  * 데이터 스트림 문제를 “매번 정렬해서 조회”하는 문제로 처리함.

* **결정적으로 막힌 이유:**
  중앙값을 매번 새로 계산하는 게 아니라, **작은 절반 / 큰 절반을 나눠서 중앙값 후보를 계속 유지해야 한다는 불변식**을 못 세웠다.

## 3. 실패 유형 / 등급

* **주 실패 유형:**
  상태·불변식 설계 실패

* **등급:**
  C 해설 후 백지 구현 가능

* **근거:**
  중앙값 계산 자체는 이해했지만, `addNum()`과 `findMedian()`이 반복 호출되는 클래스 설계 문제로 보지 못했다. 핵심 자료구조인 **두 개의 heap**과 그 사이의 균형 조건을 해설로 배워야 풀 수 있는 상태였다.

## 4. 일반화해서 공부할 요소

* **알고리즘/자료구조 신호:**
  `stream`, `add`, `find`, `median`, `repeated call`이 나오면 매번 정렬하지 말고 **상태를 유지하는 자료구조**를 의심해야 함.

* **상태/불변식:**

  ```text
  작은 절반 max heap
  큰 절반 min heap
  두 heap 크기 차이는 최대 1
  작은 절반의 모든 값 <= 큰 절반의 모든 값
  ```

* **파이썬 문법/구현 포인트:**
  Python `heapq`는 min heap만 기본 제공하므로 max heap은 음수로 저장한다.

  ```python
  heapq.heappush(max_heap, -num)
  value = -heapq.heappop(max_heap)
  ```

* **시간복잡도:**
  네 접근:

  ```text
  addNum: O(log n)
  findMedian: O(n log n)
  ```

  필요한 접근:

  ```text
  addNum: O(log n)
  findMedian: O(1)
  ```

* **경계조건:**

  * 데이터가 1개일 때
  * 데이터 개수가 홀수일 때
  * 데이터 개수가 짝수일 때
  * 두 heap의 크기 균형이 깨졌을 때
  * 음수, 중복값이 들어올 때

````python
import heapq

class MedianFinder:

    def __init__(self):
        self.data = []
        

    def addNum(self, num: int) -> None:
        heapq.heappush(self.data, num)
        

    def findMedian(self) -> float:
        temp_heap = self.data.copy()
        median = []
        while len(temp_heap) != 0:
            temp = heapq.heappop(temp_heap)
            median.append(temp)

        if len(median)%2 == 0:
            i = len(median)//2
            result = (median[i]+median[i-1])/2
            return result

        else:
            i = len(median)//2
            result = median[i]
            return result
# Your MedianFinder object will be instantiated and called as such:
# obj = MedianFinder()
# obj.addNum(num)
# param_2 = obj.findMedian()
````
