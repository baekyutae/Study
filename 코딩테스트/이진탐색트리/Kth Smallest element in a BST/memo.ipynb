## 1. 한 줄 진단

BST의 핵심 불변식인 **“중위 순회하면 오름차순”**을 코드 구조로 연결하지 못했고, 재귀 호출마다 `array`를 초기화해서 상태 유지가 깨졌다.

---

## 2. 내 사고가 막힌 지점

* **처음 접근:**
  모든 `node.val`을 배열에 담고 정렬한 뒤 `array[k - 1]` 반환.

* **맞았던 생각:**

  * `K번째로 작은 값`이라는 문제 해석은 맞음.
  * `val을 모아서 정렬하면 풀린다`는 브루트포스 접근은 맞음.
  * `BST 특성 활용`을 떠올린 것도 맞음.
  * `k - 1` 인덱싱도 맞음.

* **틀렸던 생각:**

  * `self.array = []`를 재귀 함수 안에 둬서 호출마다 배열이 리셋됨.
  * `root → left → right` 순서로 넣고 있어서 BST의 정렬 특성을 못 씀.
  * `find_k()`를 만들었지만 실제 `kthSmallest()` 반환 흐름에 연결하지 못함.
  * 최종 반환값이 `int`가 아니라 `array`임.

* **결정적으로 막힌 이유:**
  “BST 특성 활용”이라는 말은 떠올렸지만, 그 특성이 코드에서는 **중위 순회 left → root → right**로 표현된다는 연결이 안 됐다.

---

## 3. 실패 유형 / 등급

* **주 실패 유형:**
  상태·불변식 설계 실패

* **등급:**
  B 힌트 후 해결

* **근거:**
  문제 해석과 브루트포스 방향은 맞았고, BST를 활용해야 한다는 신호도 잡았다. 다만 `중위 순회 = 오름차순` 불변식과 `array 상태를 어디서 유지할지`에서 막혔다.

---

## 4. 일반화해서 공부할 요소

* **알고리즘/자료구조 신호:**
  BST에서 “작은 순서”, “정렬된 순서”, “k번째 작은 값”이 나오면 중위 순회를 먼저 의심.

* **상태/불변식:**
  중위 순회 중 `array`에는 항상 지금까지 방문한 값들이 오름차순으로 쌓인다.

* **파이썬 문법/구현 포인트:**
  재귀 helper를 따로 만들고, `array`는 바깥 함수에서 한 번만 생성한다.

* **시간복잡도:**
  전체 값 수집 후 정렬: `O(n log n)`
  BST 중위 순회로 수집: `O(n)`

* **경계조건:**
  `root is None`일 때 helper는 그냥 `return`해야 함.
  문제 조건상 `k`는 유효하다고 보면 됨.

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def kthSmallest(self, root: Optional[TreeNode], k: int) -> int:
        """
        K번째로 작은 node.val을 찾아라

        걍 node.val 싹다 담아서
        오름차순 정렬
        k번째 val 반환

        bst 특성 활용
        """
        self.array = []
        if root is None:
            return

        self.array.append(root.val)

        left = self.kthSmallest(root.left, k)
        right = self.kthSmallest(root.right, k)

        return self.array

    def find_k(self, array, k):
        array.sort()
        return array[k - 1]
```
