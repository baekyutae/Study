```python
class Solution:
    def simplifyPath(self, path: str) -> str:
        # 슬래시를 기준으로 경로명과 . , .. 들만 꺼냄
        # //같은 복수의 슬래시로 인해 생긴 공백은 무시
        # .면 현 상태 유지 ..면 가장 마지막 경로명을 삭제
        # 결과물을 가지고 / + 각 경로명사이에 /를 넣어 반환

        stack = path.split("/")
        answer = []
        for i in stack:
            if i == '':
                continue

            if i not in {".",".."}:
                answer.append(i)
                print(answer)
            if i == ".." and len(answer)>=1:
                answer.pop()

        return "/" + "/".join(answer)
```
