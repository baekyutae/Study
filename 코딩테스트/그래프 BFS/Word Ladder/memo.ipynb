입력란이 비어있어서, 방금 같이 본 **Word Ladder** 기준으로 정리하면 이렇게 가면 됨.

문제:
Word Ladder

패턴:
그래프 최단거리 + BFS

트리거:
한 상태에서 다른 상태로 이동하는데, **가장 짧은 변환 과정**을 구해야 한다.
각 이동 비용이 동일하다. 여기서는 “한 글자 변경”이 이동 1회.

핵심 발상:
단어를 노드로 보고, 한 글자만 바꿔서 만들 수 있는 단어를 간선으로 본 뒤 BFS로 최단 경로를 찾는다.

시작점:
`beginWord`에서 시작한다.
단, 반환값은 변환 횟수가 아니라 **변환 과정에 포함된 단어 개수**라서 시작 길이는 1이다.

자료구조:
`set`: 후보 단어가 `wordList`에 있는지 빠르게 확인하기 위해 사용한다.
`deque`: BFS 큐로 사용한다.
`visited`: 이미 방문한 단어를 다시 방문하지 않기 위해 사용한다.

방문/상태 처리:
현재 단어의 각 위치를 `a`~`z`로 바꿔 후보 단어를 만든다.
후보 단어가 `word_set`에 있고 아직 방문하지 않았다면 큐에 넣는다.
큐에는 현재 단어와 현재까지의 경로 길이를 함께 저장한다.
`endWord`를 처음 만나는 순간 그 길이가 최단거리다.

시간복잡도:
`N = wordList 길이`, `L = 단어 길이`일 때, 각 단어를 최대 한 번 방문하고 단어마다 `L * 26`개의 후보를 만들므로 개념상 `O(N * L)`, 문자열 생성 비용까지 보면 `O(N * L^2)`이다.

반드시 외울 것:

* “최단 변환 과정”이면 BFS를 떠올린다.
* 단어를 노드, 한 글자 변환 가능 관계를 간선으로 본다.
* `wordList`는 중복 제거가 아니라 빠른 탐색을 위해 `set`으로 바꾼다.
* `visited`가 없으면 같은 단어를 반복 방문할 수 있다.
* `endWord`가 `wordList`에 있어도 연결 경로가 없을 수 있으므로 마지막 `return 0`이 필요하다.

외울 필요 없는 것:

* 정답 코드 전체
* 변수명
* 예제 입력 모양
* 슬라이싱 문법 자체
* 알파벳 반복문의 정확한 작성 순서

내 사고 교정:
내가 처음 떠올린 접근:
단어의 글자를 하나씩 바꿔가며 `wordList`에 있는지 확인하고, 있으면 변환한다.

맞았던 생각:
`endWord`가 없으면 실패, 한 글자씩 바꿔 후보를 만든다, 후보가 `wordList`에 있어야 이동 가능하다는 생각은 맞았다.

틀렸던 생각:
가능한 후보 중 하나를 고르며 진행하면 최단거리를 보장할 수 없다.

다음에는 이렇게 생각하기:
“하나를 골라서 가는 것”이 아니라, 현재 거리에서 갈 수 있는 모든 후보를 큐에 넣고 레벨 단위로 탐색해야 한다.
