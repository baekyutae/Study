  2. 내 사고 흐름 요약
     처음에는 모든 칸을 전수조사해야 한다고 판단했고, 200 * 200이라 O(mn) 스캔은 가능하다고 봤다.
     0을 만나면 같은 행과 열을 바로 0으로 덮어쓰는 방식으로 시작했다.
     그 결과 새로 만든 0이 다시 원본 0처럼 인식되어 전파되는 문제를 발견했다.
     이후 -1을 임시 마커로 쓰는 방향으로 수정했지만, 원래 0이었던 칸을 덮어 원본 정보를 잃거나, 입력에 -1이 있을 수 있다는 충돌 문제에 막혔다.
  3. 막힌 이유 분석

  첫 오판:
  0을 발견한 즉시 행/열을 바꿔도 된다고 본 것.

  결정적으로 부족했던 점:
  “판단 기준은 원본 matrix의 0이어야 한다”는 불변식을 분리하지 못했다. 즉, 탐색 단계와 변경 단계를 분리해야 했는데, 한 번의 순회 안에서 읽기와 쓰기를 섞
  었다.

  왜 여기서 막혔는지:
  matrix를 동시에 세 가지 용도로 썼다. 원본 데이터, 방문 중인 판단 기준, 나중에 0으로 바꿀 표시 공간. 이 세 책임이 섞이면서 새로 만든 값이 다시 조건 판단
  에 영향을 줬고, -1 마커도 입력값과 충돌했다.

  4. 실패 유형 분류

  주 실패 유형:
  4. 상태/불변식 설계 실패

  부 실패 유형:
  5. 구현 실패

  근거:
  접근의 큰 방향, 즉 “원본 0이 있는 행/열을 기억한 뒤 나중에 바꿔야 한다”는 쪽으로는 점점 이동했다. 하지만 어떤 값을 유지해야 안전한지, 그 값이 입력값과
  충돌하지 않는지, 원본 0을 어떻게 보존할지에 대한 상태 설계가 부족했다. 구현 자체도 -1 센티널에 의존하면서 문제 조건을 견디지 못했다.

  5. 등급 판정

  판정: C에 가까운 D

  이유:
  첫 시도 실패 원인을 스스로 발견했고, 임시 마커라는 중간 아이디어까지 갔다. 하지만 최종적으로 “행/열 마킹을 별도 상태로 보관한다” 또는 “첫 행/첫 열을 마
  커로 재사용한다”는 안정적인 불변식까지는 자력으로 닿지 못했다.

  유사 유형 경험 점검:
  메모만 보면 과거 유사 유형을 연결한 흔적은 없다. 처음 만난 신호는 “배열을 순회하면서 동시에 값을 바꾸면 이후 판단이 오염된다”는 점이다. 이 신호는 보통
  2-pass 처리, 별도 마킹 배열/set, 또는 원본 공간 일부를 마커로 쓰는 문제로 연결해야 한다.

  6. 정답 풀이에서 뽑아낼 일반화 포인트

  자료구조/알고리즘 포인트:
  이 문제는 단순 시뮬레이션처럼 보이지만 핵심은 마킹이다. 원본 0의 위치를 직접 처리하지 말고, 먼저 0이 있는 row, col을 기록해야 한다. 가장 쉬운 방식은
  set 두 개를 써서 zero_rows, zero_cols를 저장하는 것이다. 공간 최적화를 요구하면 첫 행과 첫 열을 마커 배열처럼 재사용한다.

  상태/불변식 포인트:
  유지해야 하는 값은 “원본에서 0이 있었던 행 집합”과 “원본에서 0이 있었던 열 집합”이다. 이 두 정보만 있으면 어떤 칸 (r, c)가 0이 되어야 하는지는 r in
  zero_rows or c in zero_cols로 충분히 판단할 수 있다.

  복잡도/경계조건 포인트:
  전체 칸을 한 번은 봐야 하므로 O(mn)은 자연스럽다. m, n <= 200이면 충분히 가능하다. 주의할 점은 입력값에 -1 같은 임의의 값이 포함될 수 있으므로 센티널
  값을 함부로 정하면 안 된다는 것, 그리고 원본 0을 처리 중 덮어버리면 그 0이 만들어야 할 행/열 전파를 잃을 수 있다는 점이다.

  일반화 문장:
  순회 중 값 변경이 이후 조건 판단에 영향을 주면, 원본 판단 단계와 변경 단계를 분리하거나 안전한 마커 상태를 따로 설계한다.

  7. 다음에 써먹을 트리거 문장

  배열을 순회하면서 값 변경 + 변경된 값이 다시 조건에 걸림 -> 2-pass 마킹 점검

  원본 위치에 따라 전체 행/열/구간을 바꿈 -> row/col set 또는 marker 배열 점검

  임시값 -1, INF, None 같은 센티널을 쓰고 싶음 -> 입력 범위와 충돌 가능성 먼저 점검

  8. 개선 액션

  오늘 바로 할 것 1개:
  set 두 개를 쓰는 O(m+n) 공간 풀이를 백지에서 다시 구현해라. 핵심은 첫 번째 순회에서는 절대 matrix를 바꾸지 않는 것이다.

  내일 복습할 것 1개:
  첫 행/첫 열을 마커로 쓰는 O(1) 공간 풀이를 다시 구현해라. 특히 첫 행 자체에 0이 있었는지, 첫 열 자체에 0이 있었는지를 별도 boolean으로 저장하는 이유를
  설명해봐라.

  비슷한 문제에서 확인할 포인트 1개:
  “내가 지금 읽고 있는 값이 원본인가, 처리 중 만든 값인가?”를 코드 작성 전에 반드시 확인해라.

2. 0이라면? 행과 열을 모두 0으로 덮어씌우는 기능

같은 행의 좌표를 순회하며 0으로 만들고
같은 열의 좌표를 순회하며 0으로 만들고

```python
class Solution:
    def setZeroes(self, matrix) -> None:
        """
        Do not return anything, modify matrix in-place instead.
        """
        m = len(matrix)
        n = len(matrix[0])

        for row in range(m):
            for col in range(n):
                if matrix[row][col] == 0:
                    print(row, col)
                    # 열에 속한 좌표를 전부 0으로
                    for i in range(n):
                        matrix[row][i] = 0
                        # print(matrix)
                    # 행에 속한 좌표를 전부 0으로
                    for i in range(m):
                        matrix[i][col] = 0

matrix = [[1, 1, 1], [1, 0, 1], [1, 1, 1]]
test = Solution()
test.setZeroes(matrix)
print(matrix)
'''
=> 0으로 변환한건 건드리면 안되는데 0으로 변환한걸 또 0으로 변환해서 0이 되는 경우가 생김
'''
```
