1. **한 줄 진단**

* 핵심 실패 원인 1문장
  **2D 좌표 `(row, col)`를 직접 이진 탐색하려 했지만, 이 문제의 정렬 기준은 좌표 2개가 아니라 “1D로 펼쳤을 때의 순서”였다는 점을 놓쳤음.**

---

2. **내 사고가 막힌 지점**

* 처음 접근:
  `left_point = (0, 0)`, `right_point = (마지막 행, 마지막 열)`로 두고, 중간 좌표 `mid_point`를 잡아서 이진 탐색하려 함.

* 맞았던 생각:
  이 문제는 전체적으로 정렬되어 있으므로 **binary search를 쓸 수 있다**는 판단은 맞았음.
  `mid`를 잡고 `target`과 비교해서 탐색 범위를 줄이려는 방향도 맞았음.

* 틀렸던 생각:
  `number > target`이면 `row`와 `col`을 둘 다 줄이고,
  `number < target`이면 `row`와 `col`을 둘 다 늘리면 된다고 생각한 부분.

  예를 들어:

  ```python
  right_point[0] = mid_point[0] - 1
  right_point[1] = mid_point[1] - 1
  ```

  이건 “왼쪽 위 방향만 남긴다”는 뜻인데, 실제 정답은 **왼쪽 위가 아니라 같은 행 오른쪽이나 이전 행 오른쪽 끝**에 있을 수 있음.

* 결정적으로 막힌 이유:
  이 문제의 정렬은 이런 식의 2D 좌표 정렬이 아님.

  ```text
  row도 작고 col도 작은 쪽만 더 작다
  row도 크고 col도 큰 쪽만 더 크다
  ```

  가 아니라,

  ```text
  1D 배열로 펼쳤을 때 앞에 있으면 작고, 뒤에 있으면 크다
  ```

  임.

  그래서 좌표를 직접 줄이면 안 되고, **가상의 1D index `0 ~ m*n-1`을 기준으로 이진 탐색**해야 했음.

---

3. **실패 유형 / 등급**

* 주 실패 유형:
  **상태·불변식 설계 실패**

* 등급:
  **B 힌트 후 해결**

* 근거:
  binary search를 써야 한다는 큰 방향은 잡았지만, 탐색 범위를 유지하는 기준이 잘못됐음.
  힌트로 `mid // n`, `mid % n`이라는 1D → 2D 변환 공식을 받은 뒤에는 스스로 정답 코드로 전환했으므로 B가 맞음.

---

4. **일반화해서 공부할 요소**

* 알고리즘/자료구조 신호:
  `각 row가 정렬됨 + 다음 row의 첫 값이 이전 row의 마지막 값보다 큼`
  → **2D matrix지만 사실상 정렬된 1D 배열**

* 상태/불변식:
  이진 탐색에서는 항상 이 불변식을 잡아야 함.

  ```text
  left ~ right 사이에만 target 가능성이 남아 있다
  ```

  이 문제에서는 그 `left`, `right`가 좌표가 아니라 **1D index**여야 함.

* 파이썬 문법/구현 포인트:
  기존 코드에는 구현상 문제도 있음.

  ```python
  len(self.matrix)
  ```

  `self.matrix`는 정의한 적이 없으므로 `matrix`를 써야 함.

  그리고:

  ```python
  right_point[0] = ...
  ```

  `right_point`는 tuple이라 원소 수정이 안 됨.

* 시간복잡도:
  전체 원소 수가 `m * n`개이므로 이진 탐색 시간복잡도는:

  ```text
  O(log(mn))
  ```

* 경계조건:
  `right = len(matrix) * len(matrix[0]) - 1`
  `while left <= right`
  `row = mid // n`
  `col = mid % n`

  여기서 `n`은 **열 개수**.

5. **다음 개선 방향**

* 다음에 비슷한 문제를 만나면 먼저 확인할 것:
  “이 2D 구조가 진짜 2D 탐색 문제인지, 아니면 1D로 펼칠 수 있는 정렬 배열인지”를 먼저 봐야 함.

* 반복해서 연습할 포인트:
  2D matrix에서 index 변환:

  ```python
  row = index // col_count
  col = index % col_count
  ```

* 한 줄 복습 메모:
  **Search a 2D Matrix는 좌표 이진 탐색이 아니라, `m*n` 길이의 가상 1D 배열에 대한 이진 탐색이다.**

From training data.


````python
# 틀린 풀이

'''
에러 원인:
self.matrix가 아니라 matrix를 써야 함

현재 접근의 진짜 문제:
(row, col)을 동시에 움직이면 탐색 범위를 대각선으로 줄여서 값을 스킵함

힌트:
2D matrix를 정렬된 1D 배열처럼 보고 binary search 해라
mid를 row, col로 바꾸는 공식은 mid // n, mid % n

'''
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        left_point = (0,0)
        right_point = (len(self.matrix)-1, len(self.matrix[0])-1)

        while left_point[0] <= right_point[0] and left_point[1] <= right_point[1]:

            mid_point = ((left_point[0] + right_point[0])//2 , (left_point[1] + right_point[1])//2)

            if self.matrix[mid_point[0]][mid_point[1]] == target:
                return True

            elif self.matrix[mid_point[0]][mid_point[1]] > target:
                right_point[0] = mid_point[0]-1
                right_point[1] = mid_point[1]-1

            else:
                left_point[0] = mid_point[0] + 1
                left_point[1] = mid_point[1] + 1

        return False               
        
````
