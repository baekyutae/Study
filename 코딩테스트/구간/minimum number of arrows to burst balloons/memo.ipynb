1. 한 줄 진단

2. 내 사고 흐름 요약

* 문제를 보고 `start <= x <= end` 안에 화살이 들어가면 풍선이 터진다는 조건은 제대로 잡았다.
* 화살을 최소화하려면 여러 풍선이 겹치는 구간을 한 번에 처리해야 한다고 판단했다.
* 그래서 `start, end` 기준으로 정렬한 뒤, 겹치는 범위를 하나의 덩어리로 묶으려 했다.
* 하지만 `points[i][1]`이 다음 구간 안에 있는지 확인하는 식으로 접근하다가, “겹치는 묶음의 기준값”을 어떻게 갱신해야 하는지에서 막혔다.

3. 막힌 이유 분석

* 첫 오판:
  겹침 여부를 `points[i][1]`과 `points[i+1]`만 비교하려 한 점이다. 구간 문제에서는 단순히 이전 구간의 `end`가 다음 구간 안에 있는지만 보면 부족하다. 예를 들어 `[1,10]`, `[2,3]`은 겹치지만 `10`은 `[2,3]` 안에 없기 때문에 네 조건으로는 겹치지 않는다고 오판할 수 있다.

* 결정적으로 부족했던 점:
  “현재 화살 하나로 처리 가능한 공통 구간”을 상태로 유지해야 한다는 생각이 부족했다. 이 문제는 풍선을 실제로 병합하는 문제가 아니라, **하나의 x좌표가 여러 구간에 동시에 포함될 수 있는지**를 판단하는 문제다.

* 왜 여기서 막혔는지:
  네 접근은 `Merge Intervals`처럼 “구간을 덩어리로 묶는 문제”로 보고 있었다. 그런데 이 문제는 구간을 합치는 게 아니라, 각 겹침 그룹마다 **화살을 쏠 대표 위치 하나를 정하는 문제**다. 그래서 stack에 구간을 넣는 방향보다, `arrow_pos` 또는 `common_end` 같은 불변식을 세웠어야 한다.
