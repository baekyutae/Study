방문 여부를 체크할 set 자료형 visited 생성
상하좌우를 체크할 directions 생성
섬 개수 count = 0

for 모든 행 r:
    for 모든 열 c:
        if grid[r][c] == "1" 이고 아직 방문하지 않았다면:
            count += 1   # 새로운 섬 발견
            visited에 (r, c) 추가
            queue에 (r, c) 추가

            while queue가 빌 때까지:
                현재 좌표 cur_r, cur_c를 queue에서 꺼냄

                for 상하좌우 방향 x, y:
                    다음 좌표 next_r, next_c 계산

                    if 다음 좌표가 grid 범위 안이고
                       grid[next_r][next_c] == "1" 이고
                       아직 방문하지 않았다면:
                           visited에 추가
                           queue에 추가

return count
