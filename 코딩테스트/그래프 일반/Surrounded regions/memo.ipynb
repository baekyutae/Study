# 틀린 사유

## Surrounded Regions 오답 정리

1. **캡처 조건을 개별 좌표 기준으로 판단한 것이 틀림**

* `현재 좌표가 O이고 가장자리가 아니면 캡처 가능`이라고 생각했지만 틀림.
* 이 문제는 개별 칸이 아니라 **연결된 O 영역 전체가 border와 연결되어 있는지**가 핵심이다.
* 내부에 있는 O라도 border의 O와 연결되어 있으면 캡처하면 안 된다.

2. **캡처할 O를 직접 찾으려 한 방향이 틀림**

* 정답 방향은 `캡처할 O 찾기`가 아니라 `캡처하면 안 되는 O 찾기`이다.
* border에 있는 O에서 BFS/DFS를 시작해서 연결된 O들을 safe 처리해야 한다.
* 이후 남은 O만 X로 바꾼다.

3. **함수 이름 오타**

* `capture_check()`로 호출했는데 실제 정의는 `caputre_check()`였다.
* 함수명이 달라서 정상 실행되지 않는다.

4. **함수 정의 위치 문제**

* `capture_check()`를 사용한 뒤에 함수를 정의했다.
* Python에서는 함수 정의문이 실행된 뒤에 호출해야 한다.

5. **if문 줄바꿈 문법 오류**

* `and` 뒤에서 줄바꿈했는데 괄호가 없어서 문법 오류가 난다.
* 여러 조건을 줄바꿈할 때는 `if (...)` 형태로 감싸야 한다.

6. **visited.add 사용법 오류**

* `visited.add(next_r,next_c)`는 틀림.
* `set.add()`는 인자 하나만 받으므로 좌표 튜플로 넣어야 한다.
* 올바른 형태: `visited.add((next_r, next_c))`

7. **값 변경이 아니라 비교를 함**

* `board[x][y] == "X"`는 대입이 아니라 비교 연산이다.
* 값을 바꾸려면 `board[x][y] = "X"`로 써야 한다.

## 핵심 실패 원인

`O 하나가 캡처 가능한지`를 판단하려 했고, `O 영역 전체가 border와 연결되어 있는지`를 기준으로 보지 못했다.
