이 문제는 그래프에서 source 노드에서 destination 노드까지 이동 가능한 경로가 있는지 확인하는 문제다.

우선 각 노드가 어떤 노드와 연결되어 있는지 빠르게 확인하기 위해 인접 리스트를 만든다.

노드는 0번부터 n-1번까지 존재하므로, 노드 개수 n만큼 빈 리스트를 만든다.

```python
graph = [[] for _ in range(n)]
```

그리고 edges를 순회하면서 연결 정보를 graph에 저장한다.

```python
for a, b in edges:
    graph[a].append(b)
    graph[b].append(a)
```

이 문제의 간선은 양방향이므로, a에서 b로 갈 수 있을 뿐만 아니라 b에서도 a로 갈 수 있다.
그래서 양쪽에 모두 추가한다.

예를 들어 `[0, 1]`이라는 간선이 있으면 `0번 노드는 1번과 연결되어 있고`, `1번 노드는 0번과 연결되어 있다`는 뜻이다.

그다음 BFS로 source부터 탐색한다.

BFS는 큐를 사용해서 현재 갈 수 있는 노드들을 하나씩 꺼내며 탐색하는 방식이다.
이미 방문한 노드를 다시 방문하면 무한 반복이 생길 수 있으므로 visited set을 사용한다.

```python
queue = deque([source])
visited = set([source])
```

이제 queue가 빌 때까지 반복한다.

```python
while queue:
    node = queue.popleft()

    if node == destination:
        return True

    for next_node in graph[node]:
        if next_node not in visited:
            visited.add(next_node)
            queue.append(next_node)
```

현재 꺼낸 node가 destination이면 경로를 찾은 것이므로 True를 반환한다.

아직 destination이 아니라면, 현재 node와 연결된 이웃 노드들을 확인한다.
그중 아직 방문하지 않은 노드가 있으면 visited에 추가하고 queue에 넣는다.

이 과정을 반복했는데도 destination을 만나지 못했다면, source에서 destination까지 갈 수 있는 경로가 없다는 뜻이므로 False를 반환한다.


## Valid Path BFS 풀이 시간복잡도 정리

노드 개수를 `n`, 간선 개수를 `e = len(edges)`라고 하면 전체 시간복잡도는 `O(n + e)`이다.

### 1. 그래프 초기화

```python
graph = [[] for _ in range(n)]
```

노드 개수 `n`만큼 빈 리스트를 만들기 때문에 `O(n)`이 걸린다.

### 2. 간선 정보로 그래프 만들기

```python
for a, b in edges:
    graph[a].append(b)
    graph[b].append(a)
```

`edges` 배열을 한 번 순회하므로 `O(e)`가 걸린다.

무방향 그래프이기 때문에 간선 하나를 양쪽에 저장한다.

```python
graph[a].append(b)
graph[b].append(a)
```

그래서 실제 인접 리스트에 저장되는 총 원소 수는 `2e`지만, 빅오에서는 상수를 제거하므로 `O(e)`로 본다.

따라서 그래프 생성 시간은:

```text
O(n + e)
```

### 3. BFS 탐색

```python
while queue:
    node = queue.popleft()

    for next_node in graph[node]:
        ...
```

`visited`를 사용하므로 각 노드는 큐에 최대 한 번만 들어간다.
따라서 노드를 꺼내서 처리하는 비용은 최대 `O(n)`이다.

그리고 `for next_node in graph[node]`는 각 노드의 연결된 노드를 확인한다.

모든 노드의 인접 리스트 길이를 전부 합치면 무방향 그래프에서는 `2e`이다.

```text
모든 인접 리스트 길이의 합 = 2e
```

따라서 이웃 노드를 확인하는 전체 비용은 `O(e)`이다.

그래서 BFS 탐색 시간은:

```text
O(n + e)
```

### 4. 최종 시간복잡도

```text
그래프 생성: O(n + e)
BFS 탐색: O(n + e)

전체 시간복잡도: O(n + e)
```

LeetCode식으로 쓰면:

```text
O(n + m)
```

여기서 `m = len(edges)`이다.

핵심은 `graph 길이 * 노드별 연결 개수`로 보는 게 아니라,
`모든 노드의 인접 리스트 길이 합 = 2 * edges.length`로 본다는 것이다.

따라서 최종 시간복잡도는:

```text
O(n + edges.length)
```
