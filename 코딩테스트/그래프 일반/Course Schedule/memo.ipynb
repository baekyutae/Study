문제:
    Course Schedule

패턴:
    Graph / 위상정렬 / BFS

트리거:
    - 선수 조건이 있다.
    - 어떤 작업을 하기 전에 먼저 끝내야 하는 작업이 있다.
    - 모든 작업을 끝낼 수 있는지 묻는다.
    - 순서 관계가 꼬이면 불가능하다.

핵심 발상:
    각 과목을 듣기 위해 필요한 선수과목 개수(indegree)를 센다.
    선수과목을 다 들은 과목, 즉 indegree가 0인 과목부터 큐에 넣고 수강한다.
    과목 하나를 수강할 때마다 그 과목을 필요로 하던 후행 과목들의 indegree를 1씩 줄인다.
    최종적으로 수강 완료한 과목 수가 전체 과목 수와 같으면 True다.
    끝까지 처리하지 못한 과목이 남으면 사이클이 있다는 뜻이므로 False다.

시작점:
    indegree가 0인 과목들
    = 선수과목 없이 바로 들을 수 있는 과목들

자료구조:
    graph:
        graph[선수과목] = [그 선수과목을 필요로 하는 후행 과목들]

    indegree:
        indegree[과목] = 해당 과목을 듣기 위해 아직 필요한 선수과목 개수

    queue:
        현재 수강 가능한 과목들, 즉 indegree가 0인 과목들 저장

    finished:
        수강 완료한 과목 수

방문/상태 처리:
    별도의 visited는 사용하지 않는다.
    대신 indegree를 줄이면서 상태를 관리한다.

    indegree가 0이 되는 순간:
        필요한 선수과목을 모두 들었다는 뜻이므로 큐에 넣는다.

    큐에서 꺼내는 순간:
        해당 과목을 수강 완료한 것으로 처리하고 finished += 1 한다.

시간복잡도:
    O(V + E)

    V = 과목 수
    E = 선수과목 관계 수

    모든 과목과 모든 선수과목 관계를 한 번씩 처리하기 때문이다.

공간복잡도:
    O(V + E)

    graph에 간선 정보를 저장하고,
    indegree와 queue에 과목 정보를 저장하기 때문이다.

다음에 떠올릴 문장:
    “선수 조건이 있는 작업 순서 문제는 위상정렬을 의심한다.
    indegree가 0인 것부터 처리하고,
    처리할 때마다 후행 노드의 indegree를 줄인다.
    끝까지 다 처리하면 가능, 남으면 사이클이다.”
